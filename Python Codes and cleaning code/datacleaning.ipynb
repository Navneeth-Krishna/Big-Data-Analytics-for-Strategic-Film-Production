```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Load dataset
dataset = pd.read_csv(r"C:\Users\Vishal\Downloads\dash.csv")

# Data Exploration
print("Dataset Head:")
print(dataset.head())

print("\nDataset Info:")
print(dataset.info())

# Check for missing values
print("\nMissing Values:")
print(dataset.isnull().sum())

# Fill missing values for 'Genre' with the mode
genre_mode = dataset['Genre'].mode()[0]
dataset['Genre'].fillna(genre_mode, inplace=True)

# Fill missing values for 'Production Budget ($)' with the median
budget_median = dataset['Production Budget ($)'].median()
dataset['Production Budget ($)'].fillna(budget_median, inplace=True)

# Fill missing values for 'Trailer' with 'No Trailer'
dataset['Trailer'].fillna('No Trailer', inplace=True)

# Check again for missing values to ensure they have been handled
print("\nMissing Values After Cleaning:")
print(dataset.isnull().sum())

# Verify changes
print("\nDataset Head After Cleaning:")
print(dataset.head())

# Save the cleaned dataset to a new CSV file
cleaned_file_path = r"C:\Users\Vishal\Downloads\cleaned_dash.csv"
dataset.to_csv(cleaned_file_path, index=False)

print(f"Cleaned dataset saved to {cleaned_file_path}")
```
