```python
from bs4 import BeautifulSoup
import requests
import csv
from os.path import isdir
from os import mkdir

url = 'https://www.the-numbers.com/United-States/movies/year/2020'
output_folder = 'outputs/thenumbers'

if not isdir('outputs'):
    mkdir('outputs')

if not isdir(output_folder):
    mkdir(output_folder)

header = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/121.0.0.0 Safari/537.36 Edg/121.0.0.0',
}
page = requests.get(url, headers=header)
soup = BeautifulSoup(page.text, 'html.parser')

table = soup.find_all('table')[0]

rows = table.find_all('tr')

table_content = rows[1:]

header_row = ['Movie', 'Co-Producing Countries', 'Genre', 'Production Budget ($)', 'Worldwide Box Office to Date ($)', 'Trailer']

with open(f"{output_folder}/thenumbers_usa_2020.csv", 'w', newline='', encoding='utf-8') as csvfile:
    c = csv.writer(csvfile)
    c.writerow(header_row)
    for i in range(len(table_content)):
        cells = table_content[i].find_all('td')
        movie_name = cells[0].text
        countries = cells[1].text
        genre = cells[2].text
        try:
            budget = int(cells[3].text.strip('$').replace(',', ''))
        except ValueError:
            budget = None
        try:
            box_office = int(cells[4].text.strip('$').replace(',', ''))
        except ValueError:
            box_office = None
        trailer = None
        if cells[5].find('a') is not None:
            trailer = 'https://www.the-numbers.com' + cells[5].find('a')['href']
        c.writerow([movie_name, countries, genre, budget, box_office, trailer])

print('DONE!')
```

```python
from bs4 import BeautifulSoup
import requests
import csv
from os.path import isdir
from os import mkdir
from IPython.display import FileLink

# URL and output folder
url = 'https://www.the-numbers.com/United-States/movies/year/2020'
output_folder = 'outputs/thenumbers'

# Ensure output directories exist
if not isdir('outputs'):
    mkdir('outputs')

if not isdir(output_folder):
    mkdir(output_folder)

# HTTP headers
header = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/121.0.0.0 Safari/537.36 Edg/121.0.0.0',
}

# Fetch the page content
page = requests.get(url, headers=header)
soup = BeautifulSoup(page.text, 'html.parser')

# Find the table and rows
table = soup.find_all('table')[0]
rows = table.find_all('tr')
table_content = rows[1:]

# Define CSV header
header_row = ['Movie', 'Co-Producing Countries', 'Genre', 'Production Budget ($)', 'Worldwide Box Office to Date ($)', 'Trailer']

# Write to CSV
csv_file_path = f"{output_folder}/thenumbers_usa_2020.csv"
with open(csv_file_path, 'w', newline='', encoding='utf-8') as csvfile:
    c = csv.writer(csvfile)
    c.writerow(header_row)
    for row in table_content:
        cells = row.find_all('td')
        movie_name = cells[0].text.strip()
        countries = cells[1].text.strip()
        genre = cells[2].text.strip()
        try:
            budget = int(cells[3].text.strip('$').replace(',', ''))
        except ValueError:
            budget = None
        try:
            box_office = int(cells[4].text.strip('$').replace(',', ''))
        except ValueError:
            box_office = None
        trailer = None
        if cells[5].find('a') is not None:
            trailer = 'https://www.the-numbers.com' + cells[5].find('a')['href']
        c.writerow([movie_name, countries, genre, budget, box_office, trailer])

print('DONE!')

# Provide download link if running in a notebook
csv_file_path
```
